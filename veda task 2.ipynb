```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=30)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
print("Observation:")
print("The histogram shows the distribution of sales values. Most orders have relatively low sales, while a smaller number of orders have very high sales.")
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df["Profit"])

plt.title("Distribution of Profit")
plt.ylabel("Profit")

plt.show()
```

```python
print("Observation:")
print("The boxplot shows the spread of profit values and helps identify unusually high or low profit values that may be outliers.")
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Category", y="Sales", estimator="sum", errorbar=None)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
print("Observation:")
print("The bar chart compares total sales across the different product categories and shows which category contributes the most to overall sales.")
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Sales", y="Profit")

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
print("Observation:")
print("The scatter plot shows the relationship between sales and profit. It helps identify whether higher sales are generally associated with higher profit and also highlights unusual observations.")
```

```python
plt.figure(figsize=(8, 6))

corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("Observation:")
print("The correlation heatmap shows the strength and direction of relationships between numerical variables. Values closer to +1 indicate a strong positive relationship, while values closer to -1 indicate a strong negative relationship.")
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales.index = monthly_sales.index.to_timestamp()

monthly_sales.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index, monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
print("Observation:")
print("The line chart shows how total sales change from month to month. It helps identify periods of higher and lower sales and overall sales trends over time.")
```

```python
print("TOP 3 INSIGHTS")
print()


category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print("1. Sales by Category:")
print(category_sales)

print()


category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

print("2. Profit by Category:")
print(category_profit)

print()


monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

print("3. Top 5 Months by Sales:")
print(monthly_sales.sort_values(ascending=False).head())
```
